```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('QVI_transaction_data.xlsx')
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df.shape
```

```python
df.info()
```

```python
df.columns
```

```python
df['PROD_NAME']
```

```python
df["WEIGHT"] = df["PROD_NAME"].str.extract(r'(\d+)\s*g', expand=False)
```

```python
df["PROD_NAME"] = df["PROD_NAME"].str.replace(r'\d+\s*g', '', regex=True).str.strip()
```

```python
plt.style.use('ggplot')
plt.scatter(df['LYLTY_CARD_NBR'], df['DATE'],color='red', alpha=0.5)
plt.title('Loyalty Card Number vs Date')
plt.show()
```
